```sql
%python
import os

# 📌 Configuração
tabelas_alvo = ['TAB_VENDA_ITEM', 'TAB_VENDA', 'TAB_PRODUTO']
diretorio_base = "/Volumes/workspace/default/bertoferramentas/"

catalog_name = "workspace" # Ex: "main", "hive_metastore"
schema_name = "bertoferramentas"

print("Iniciando virtualização dos dados...")

for tabela in tabelas_alvo:
    caminho_csv = os.path.join(diretorio_base, f"{tabela}.csv")
    if not os.path.exists(caminho_csv):
        caminho_csv = os.path.join(diretorio_base, f"{tabela.lower()}.csv")
        
    tabela_virtual = f"{catalog_name}.{schema_name}.{tabela.lower()}"
    
    try:
        # Lê o CSV via Spark
        df_spark = spark.read.format("csv") \
            .option("header", "true") \
            .option("inferSchema", "false") \
            .load(caminho_csv)
        
        # Salva (Virtualiza) como tabela no Catálogo
        df_spark.write.mode("overwrite").saveAsTable(tabela_virtual)
        print(f"✅ Sucesso: {tabela} registrada no catálogo como {tabela_virtual}")
        
    except Exception as e:
        print(f"❌ Erro ao virtualizar {tabela}: {e}")
```
